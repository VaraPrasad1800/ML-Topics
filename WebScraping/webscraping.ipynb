```python
import requests
import pandas as pd
from bs4 import BeautifulSoup

headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 6.3; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/80.0.3987.162 Safari/537.36'}

frames = []
for j in range(1, 300):
    url = f"https://www.ambitionbox.com/list-of-companies?page={j}"
    webpage = requests.get(url, headers=headers).text
    soup = BeautifulSoup(webpage, 'html.parser')
    companies = soup.select('div[data-testid^="loc-company-card-"]')

    names, ratings, reviews, sector, location = [], [], [], [], []

    for i in companies:
        names.append(i.find('h2').text)
        ratings.append(i.find('span', class_='text-primary-text').text)
        reviews.append(i.find('span', class_='pl-1').text)
        info = i.find('span', class_='overflow-hidden').text

        parts = info.split("|")
        sector.append(parts[0].strip())
        location.append(parts[1].strip() if len(parts) > 1 else None)

    frames.append(pd.DataFrame({
        'name': names,
        'rating': ratings,
        'review': reviews,
        'sector': sector,
        'location': location,
    }))

final = pd.concat(frames, ignore_index=True)
final.shape
```

```python
import time
import requests
import pandas as pd
from bs4 import BeautifulSoup

headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'}

session = requests.Session()
session.headers.update(headers)

frames = []
for j in range(1, 5):
    url = f"https://www.ambitionbox.com/list-of-companies?page={j}"
    resp = session.get(url, timeout=15)
    soup = BeautifulSoup(resp.text, 'html.parser')
    companies = soup.select('div[data-testid^="loc-company-card-"]')

    print(j, resp.status_code, len(companies), resp.url)   # diagnostics

    if not companies:
        print("No cards on this page. Page title:", soup.title.text if soup.title else None)
        break

    rows = []
    for i in companies:
        info = i.find('span', class_='overflow-hidden')
        parts = info.text.split("|") if info else []
        rows.append({
            'name': i.find('h2').text if i.find('h2') else None,
            'rating': i.find('span', class_='text-primary-text').text if i.find('span', class_='text-primary-text') else None,
            'review': i.find('span', class_='pl-1').text if i.find('span', class_='pl-1') else None,
            'sector': parts[0].strip() if len(parts) > 0 else None,
            'location': parts[1].strip() if len(parts) > 1 else None,
        })
    frames.append(pd.DataFrame(rows))
    time.sleep(1.5)   # be polite, avoid getting blocked

final = pd.concat(frames, ignore_index=True)
print(final.shape)
final
```
